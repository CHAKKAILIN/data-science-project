# Activity 7.01: Using Heatmaps to Find Patterns in Flight Passengers' Data

```python
# 02: Import the necessary modules and enable plotting
%matplotlib inline
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
sns.set()
```

```python
# 03: Load the flight_details.csv dataset
data = pd.read_csv("flight_details.csv")
data
```

```python
# 04: Pivot data so months are rows and years are columns
month_order = ['January', 'February', 'March', 'April', 'May', 'June',
               'July', 'August', 'September', 'October', 'November', 'December']
flight_data = data.pivot(index='Months', columns='Years', values='Passengers')
flight_data = flight_data.reindex(month_order)
flight_data
```

```python
# 05: Use a sequential colormap (lowest = brightest, highest = darkest)
plt.figure(dpi=200)
sns.heatmap(flight_data, cmap='YlGnBu')
plt.title("Flight Passengers from 1949 to 1960")
plt.xlabel("Years")
plt.ylabel("Months")
plt.show()
```
